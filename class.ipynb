```python
from pydantic import BaseModel
class User(BaseModel):
  name:str
  email:str
class AppConfig(BaseModel):
  app_type:str
  owner:User
app = AppConfig(
    app_type="crm",
    owner={
        "name":"Akanksha",
        "email":"abc@gmail.com"
    }
)
print(app.model_dump())

```

```python
from pydantic import BaseModel
class Intent(BaseModel):
  app_type:str
  features:list[str]
intent= Intent(
    app_type="crm",
    features=["login","contacts"]
)
print(intent.model_dump())
```

```python

# validate erro
from pydantic import BaseModel,ValidationError
class User(BaseModel):
  age:int
try:
  user = User(age = "hello")
except ValidationError as e:
  print(e)
```

```python
from pydantic import BaseModel
class Intent(BaseModel):
  app_type:str
  features:list[str]
  roles:list[str]
intent =Intent(
    app_type="crm",
    features=["login","contact"],
    roles=["admin","user"]

)
print(intent.model_dump())
```
